```python
from ultralytics import YOLO
import os

# ==========================================
# 1. KHỞI TẠO MÔ HÌNH
# ==========================================
# Sử dụng YOLO26s (Small) phù hợp với mục tiêu nhỏ là ô trắc nghiệm
model = YOLO("yolo26s.pt")

# ==========================================
# 2. HUẤN LUYỆN (TRAIN)
# ==========================================
# Chúng ta sẽ sử dụng file data.yaml đã tạo ở bước trước
results = model.train(
    data="YOLO_Dataset/data.yaml", # Đường dẫn đến file cấu hình dataset
    epochs=50,                    # Số epoch (có thể điều chỉnh tùy độ hội tụ)
    imgsz=960,                     # Kích thước ảnh đầu vào
    project="/content/drive/MyDrive/OMR-Datasets/train_detect/Yolo26s/OMR_Localization",    # Tên dự án để lưu kết quả
    name="yolo26s_omr_detect",     # Tên lần chạy
)

```

```python
# ==========================================
# 3. ĐÁNH GIÁ TRÊN TẬP TEST (EVALUATION)
# ==========================================
# Sau khi huấn luyện, YOLO mặc định chỉ báo cáo kết quả trên tập 'val'.
# Để đáp ứng Reviewer 1, chúng ta phải chạy đánh giá trên tập 'test' cô lập.

print("\n" + "="*50)
print("📊 ĐANG ĐÁNH GIÁ HIỆU NĂNG TRÊN TẬP TEST CÔ LẬP...")
print("="*50)

# Chạy validation trên split 'test'
metrics = model.val(split='test')

# In các chỉ số quan trọng ra màn hình
print(f"\n🎯 Kết quả trên tập Test (Unseen Data):")
print(f" - mAP50: {metrics.results_dict['metrics/mAP50(B)']:.4f}")
print(f" - mAP50-95: {metrics.results_dict['metrics/mAP50-95(B)']:.4f}")
print(f" - Precision: {metrics.results_dict['metrics/precision(B)']:.4f}")
print(f" - Recall: {metrics.results_dict['metrics/recall(B)']:.4f}")
```
